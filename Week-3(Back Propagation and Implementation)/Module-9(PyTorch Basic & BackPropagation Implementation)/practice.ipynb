```python
X = torch.tensor([[1.0, 2.0]])
W1 = torch.tensor([[0.5, 0.1], [0.2, 0.4]])
b1 = torch.tensor([[0.1, -0.1]])
W2 = torch.tensor([[0.6], [0.3]])
b2 = torch.tensor([[0.05]])

# Forward pass execution
Z1 = torch.matmul(X, W1) + b1
A1 = torch.relu(Z1)
output = torch.matmul(A1, W2) + b2

print(output)
```
