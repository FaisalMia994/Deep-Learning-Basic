# Reduction Operations

```python
e = torch.randint(size=(2,3),low=0,high=10,dtype=torch.float32)
e
```

```python
torch.sum(e)
```

```python
# row or column wise 

print(torch.sum(e,dim=0)) # column-wise

print(torch.sum(e,dim=1)) # row-wise
```

```python
torch.mean(e)
```

```python
torch.mean(e,dim=0)
```

```python
torch.median(e)
```

```python
# max and min

print(torch.max(e))
print(torch.min(e))
```

```python
torch.prod(e)
```

```python
torch.std(e)
```

```python
torch.var(e)
```

```python
# argmax and argmin

print(torch.argmax(e)) # return the index of min-val
print(torch.argmin(e)) # return the index of max-val
```

# Comparison Operations


```python
i = torch.randint(size=(2,3),low=0,high=10)

j = torch.randint(size=(2,3),low=0,high=10)

print(i)
print(j)
```

```python
# greater than , less than , equal to , not equal to

print(i > j)
print(i < j)
print(i == j)
print(i != j)
```

```python
res = torch.eq(i,j)
res
```

```python
torch.add(i,j)
```

# Special Function

```python
k = torch.tensor([0.001,20,0.0003])
```

```python
torch.log(k) # loge(k) works here
```

```python
# exponential

torch.exp(k)
```

```python
# sqrt

torch.sqrt(k)
```

```python
# sigmoid function

print(torch.sigmoid(k))

# softmax

print(torch.softmax(k,dim=0))

# relu

print(torch.relu(k))
```
