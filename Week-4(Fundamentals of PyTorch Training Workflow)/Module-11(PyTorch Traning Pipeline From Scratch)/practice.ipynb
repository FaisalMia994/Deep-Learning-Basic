```python
import torch

x = torch.tensor([2.0], requires_grad=True)
y = x**2          # Intermediate tensor (the "middle")
y.retain_grad()   # Required: tells PyTorch to save y.grad

z = y * 3         # Final output
z.backward()      # Can backpropagate from the end

print("Gradient at x (leaf):", x.grad)      # tensor([12.]) -> dy/dx = 2*x * 3 = 12
print("Gradient at y (middle):", y.grad)    # tensor([3.])  -> dz/dy = 3
```
